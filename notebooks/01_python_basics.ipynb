## 1. Variables and types

A variable is a name that refers to a value. Python works out the type for you.

Text after `#` is a **comment**. Python ignores it.

```python
n_cells = 42             # int (whole number)
pixel_size_um = 0.65     # float (decimal number)
sample = 'HeLa'          # str (text)
is_stained = True        # bool (True / False)

print(type(n_cells), type(pixel_size_um), type(sample), type(is_stained))
```

### 1.1 Arithmetic
| operator | meaning |
|---|---|
| `+ - * /` | add, subtract, multiply, divide |
| `**` | power |
| `//` | whole-number division |
| `%` | remainder |

```python
area_px = 1200
area_um2 = area_px * pixel_size_um ** 2
print('Area in square micrometres:', area_um2)

print('7 / 2  =', 7 / 2)
print('7 // 2 =', 7 // 2)
print('7 % 2  =', 7 % 2)

# Change a variable in place
n_cells = n_cells + 8
n_cells += 10            # short for n_cells = n_cells + 10
print('n_cells is now', n_cells)
```

### 1.2 Comparisons and booleans
Comparisons give `True` or `False`. Combine them with `and`, `or`, `not`.

```python
intensity = 180
print(intensity > 100)
print(intensity == 180)     # == compares, = assigns
print(intensity != 0)
print(intensity > 100 and intensity < 200)
print(not is_stained)
```

## 2. Strings

Strings hold text. **f-strings** (with an `f` before the quote) insert values into text.

```python
filename = 'nuclei_sample01_DAPI.tif'

print(len(filename))
print(filename.upper())
print(filename.endswith('.tif'))
print(filename.replace('DAPI', 'GFP'))
print(filename.split('_'))          # split into a list of parts

print(f'{sample} image with {n_cells} cells, pixel size {pixel_size_um} um')
print(f'Area: {area_um2:.1f} um^2')  # :.1f means 1 decimal place
```

## 3. Collections

### 3.1 Lists
A **list** is an ordered collection you can change.

**Note:** Python counts from **0**: the first element has index 0. A slice `start:stop` includes `start` but **excludes** `stop`.

```python
intensities = [10, 52, 87, 120, 255]
print('First element :', intensities[0])
print('Last element  :', intensities[-1])
print('Elements 1-3  :', intensities[1:4])
print('Length        :', len(intensities))
print('Min, max, sum :', min(intensities), max(intensities), sum(intensities))

intensities.append(300)
print('After append  :', intensities)
intensities[0] = 15
print('After change  :', intensities)
```

### 3.2 Tuples
A **tuple** is like a list but cannot be changed. Image shapes are tuples, for example `(512, 512)`.

```python
shape = (512, 1024)
rows, cols = shape          # 'unpacking' a tuple into two variables
print(f'{rows} rows and {cols} columns')
```

### 3.3 Dictionaries
A **dictionary** maps keys to values, which is handy for metadata.

```python
experiment = {'name': 'nuclei_01', 'channels': 3, 'pixel_size_um': 0.65}
print(experiment['name'])

experiment['date'] = '2026-11-21'     # add a new key
print(experiment)
print('Keys:', list(experiment.keys()))
print('Has a date?', 'date' in experiment)
```

## 4. Loops and conditions

Indentation (4 spaces) tells Python which lines belong inside the loop or `if` block.

```python
threshold = 100
for value in intensities:
    if value > 255:
        print(value, '-> out of 8-bit range!')
    elif value > threshold:
        print(value, '-> bright')
    else:
        print(value, '-> dark')
```

```python
# range() gives a sequence of numbers; enumerate() gives index and value together
for i in range(3):
    print('iteration', i)

for i, value in enumerate(intensities):
    print(f'index {i}: {value}')

# Loop over a dictionary
for key, value in experiment.items():
    print(f'{key:>14}: {value}')
```

```python
# A list comprehension builds a new list in one line
bright = [v for v in intensities if v > threshold]
doubled = [v * 2 for v in intensities]
print('Bright values:', bright)
print('Doubled      :', doubled)
```

## 5. Functions

Functions package code so you can reuse it. Arguments can have default values.

```python
def pixels_to_um2(area_px, pixel_size_um=0.65):
    """Convert an area in pixels to square micrometres."""
    return area_px * pixel_size_um ** 2

print(pixels_to_um2(1200))
print(pixels_to_um2(1200, pixel_size_um=0.1))
help(pixels_to_um2)
```

## 7. NumPy: arrays are images

A grayscale image is a 2D grid of numbers, one per pixel. NumPy arrays store such grids efficiently and let you compute on all pixels at once, with no loops needed.

```python
import numpy as np

a = np.array([[0, 50, 100],
              [150, 200, 250]])

print(a)
print('shape:', a.shape)   # (rows, columns)
print('ndim :', a.ndim)
print('dtype:', a.dtype)
print('size :', a.size)
```

```python
# Creating arrays
zeros = np.zeros((3, 4))
ones = np.ones((2, 2), dtype=np.uint8)
ramp = np.arange(0, 10, 2)          # 0, 2, 4, 6, 8
random = np.random.default_rng(seed=0).random((2, 3))

print(zeros, ones, ramp, random, sep='\n\n')
```

### 7.1 Indexing and slicing
Index as `array[row, column]`. Slices use `start:stop:step`.

```python
grid = np.arange(25).reshape(5, 5)
print(grid, '\n')
print('Row 1, column 3 :', grid[1, 3])
print('Whole row 2     :', grid[2, :])
print('Whole column 0  :', grid[:, 0])
print('Top-left 2x2    :\n', grid[:2, :2])
print('Every 2nd row   :\n', grid[::2])
```

### 7.2 Vectorised maths
Operations apply to every element at once. Compare this with the loop you would need for a list.

```python
print(grid * 2)
print(grid + 100)
print('mean:', grid.mean(), ' min:', grid.min(), ' max:', grid.max(), ' std:', round(grid.std(), 2))
print('mean of each column:', grid.mean(axis=0))
print('mean of each row   :', grid.mean(axis=1))
```

### 7.3 Boolean masks
A comparison on an array gives a **mask** of `True`/`False` values. Masks are the basis of image segmentation.

```python
mask = grid > 12
print(mask)
print('Values above 12:', grid[mask])
print('How many       :', mask.sum())

# Use a mask to change selected values
clipped = grid.copy()
clipped[clipped > 20] = 20
print(clipped)
```

### 7.4 A first look at an array as an image
matplotlib's `imshow` draws an array as an image. Each number becomes the brightness of one pixel.

```python
import matplotlib.pyplot as plt

fig, axes = plt.subplots(1, 2, figsize=(8, 4))
axes[0].imshow(grid, cmap='gray')
axes[0].set_title('grid')
axes[1].imshow(mask, cmap='gray')
axes[1].set_title('grid > 12')
plt.show()
```
